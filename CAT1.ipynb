```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Loading the Dataset
data = pd.read_csv('C:/Users/HILLARY/Desktop/Assignments/Machine Learning/Student Performance Dataset.csv')
print("First 5 rows:\n", data.head())

# Check Column Names and Structure
print("\nColumn Names:", data.columns)
print("\nDataset Info:\n")
print(data.info())
print("\nSummary Statistics:\n", data.describe())

# Checking for Missing Values
print("\nMissing Values:\n", data.isnull().sum())  # Check for missing values first


# Data Exploration
sns.histplot(data['overall_score'], kde=True)
plt.title('Distribution of Overall Scores')
plt.show()

# Encoding Categorical Data
encoder = LabelEncoder()
data['gender'] = encoder.fit_transform(data['gender'])
data['extracurricular'] = encoder.fit_transform(data['extracurricular'])

# Normalizing Numerical Features
features_to_scale = ['study_time', 'math_score', 'science_score', 'english_score']
scaler = StandardScaler()
data[features_to_scale] = scaler.fit_transform(data[features_to_scale])

print("\nProcessed Data Sample:\n", data.head())
```
